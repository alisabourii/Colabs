```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Created on Wed Dec 24 15:42:50 2025

@author: abdullahbalci
"""

class Runner:

    def __init__(self,name):
        self.name = name
        self.runs = []

    def add_run(self,distance,duration):
        self.runs.append([distance,duration])

    def total_distance(self):
        t_distance = 0
        for i in range(len(self.runs)):
            t_distance += self.runs[i][0]

        return t_distance

    def total_duration(self):
        t_duration = 0
        for i in range(len(self.runs)):
            t_duration += self.runs[i][1]

        return t_duration

    def average_pace(self):
        total_dist = self.total_distance()
        total_dur = self.total_duration()
        avg_pace = total_dur/total_dist
        t_sec = avg_pace*60
        pace_min = int(t_sec//60)
        pace_sec = int(t_sec%60)

        return [str(pace_min) +":" +str(pace_sec)]

    def longest_run(self):
        longest = 0
        for i in range(len(self.runs)):
            if self.runs[i][0] >longest:
                longest = self.runs[i][0]
        return longest

runner = Runner("Ali")
runner.add_run(5,25)
runner.add_run(5,30)
runner.add_run(10,55)
runner.add_run(10,67)
runner.add_run(5,23)
```
